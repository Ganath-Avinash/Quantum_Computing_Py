```python
# Q1
from qiskit import QuantumCircuit

# Create a quantum circuit with 1 qubit
qc = QuantumCircuit(1)

# Display the circuit
print(qc)
```

```python
# Q2
from qiskit import QuantumCircuit

# Create a circuit with 1 qubit
qc = QuantumCircuit(1)

# Apply Pauli-X gate
qc.x(0)

# Display the circuit
print(qc)
```

```python
# Q3
from qiskit import QuantumCircuit

# Create a circuit with 1 qubit
qc = QuantumCircuit(1)

# Apply Hadamard gate
qc.h(0)

# Display the circuit
print(qc)
```

```python
# Q1 , Q2 With Measurement
from qiskit import QuantumCircuit

# Create 1 qubit and 1 classical bit
qc = QuantumCircuit(1, 1)

# Apply Hadamard gate
qc.h(0)

# Measure qubit
qc.measure(0, 0)

# Display circuit
print(qc)
```

```python
# Set of major commands used
qc = QuantumCircuit(1)   # Create 1 qubit
qc.x(0)                  # Pauli-X / NOT gate
qc.h(0)                  # Hadamard gate
print(qc)                # Display circuit
```

```python
# Q5
from qiskit import QuantumCircuit

qc = QuantumCircuit(2)

qc.h(0)
qc.cx(0, 1)

print(qc)
```

```python
# Q6
from qiskit import QuantumCircuit
from qiskit_aer import AerSimulator

qc = QuantumCircuit(1, 1)

qc.h(0)
qc.measure(0, 0)

simulator = AerSimulator()
result = simulator.run(qc, shots=1024).result()

print(result.get_counts())
```

```python
# Q7
from qiskit import QuantumCircuit
from qiskit_aer import AerSimulator

qc = QuantumCircuit(2, 2)

qc.h(0)
qc.h(1)

qc.measure(0, 0)
qc.measure(1, 1)

simulator = AerSimulator()
result = simulator.run(qc, shots=1024).result()

print(result.get_counts())
```

```python
# Q9
from qiskit import QuantumCircuit
from qiskit_aer import AerSimulator

qc = QuantumCircuit(2, 2)

qc.h(0)
qc.cx(0, 1)

qc.measure(0, 0)
qc.measure(1, 1)

simulator = AerSimulator()

result = simulator.run(qc, shots=1024).result()
counts = result.get_counts()

print("Measurement Results:", counts)
```

```python
# Q10
from qiskit import QuantumCircuit
from qiskit_aer import AerSimulator

# Classical bit: definite value
classical = QuantumCircuit(1, 1)
classical.measure(0, 0)

# Quantum bit: superposition
quantum = QuantumCircuit(1, 1)
quantum.h(0)
quantum.measure(0, 0)

simulator = AerSimulator()

print("Classical bit:")
print(simulator.run(classical, shots=10).result().get_counts())

print("Quantum bit:")
print(simulator.run(quantum, shots=1000).result().get_counts())
```
