```python
# Q1
from qiskit import QuantumCircuit
from qiskit.visualization import plot_bloch_vector
import matplotlib.pyplot as plt

# Create a 1-qubit circuit
qc = QuantumCircuit(1)

# Display circuit
print(qc)

# Create Bloch Sphere for |0>
fig = plot_bloch_vector([0, 0, 1])

# Save the Bloch Sphere
fig.savefig("bloch_sphere.png", dpi=300, bbox_inches="tight")

# Show it
plt.show()
```

```python
# Q2
from qiskit import QuantumCircuit
from qiskit.visualization import plot_bloch_vector

# Create a 1-qubit circuit
qc = QuantumCircuit(1)

# Apply Pauli-X gate
qc.x(0)

# Display the circuit
print(qc)

# |1> corresponds to Bloch vector (0, 0, -1)
fig = plot_bloch_vector([0, 0, -1])

# Save the Bloch Sphere as a clear image
fig.savefig(
    "pauli_x_bloch_sphere.png",
    dpi=300,
    bbox_inches="tight"
)

print("Bloch Sphere saved as pauli_x_bloch_sphere.png")
```

```python
# Q3
from qiskit import QuantumCircuit
from qiskit.visualization import plot_bloch_vector

# Create a 1-qubit circuit
qc = QuantumCircuit(1)

# Apply Hadamard gate
qc.h(0)

# Display the circuit
print(qc)

# Hadamard creates the |+> superposition state
fig = plot_bloch_vector([1, 0, 0])

# Save the Bloch Sphere
fig.savefig(
    "hadamard_bloch_sphere.png",
    dpi=300,
    bbox_inches="tight"
)

print("Bloch Sphere saved as hadamard_bloch_sphere.png")
```

```python
# Q4
from qiskit import QuantumCircuit
from qiskit.visualization import plot_bloch_vector

# Create a 1-qubit circuit
qc = QuantumCircuit(1)

# Apply Pauli-Y gate
qc.y(0)

# Display the circuit
print(qc)

# After Y|0> = i|1>
# The global phase i does not affect the Bloch Sphere.
# Therefore, the state is |1>, with Bloch vector (0, 0, -1)
fig = plot_bloch_vector([0, 0, -1])

# Save the Bloch Sphere
fig.savefig(
    "pauli_y_bloch_sphere.png",
    dpi=300,
    bbox_inches="tight"
)

print("Bloch Sphere saved as pauli_y_bloch_sphere.png")
```

```python
# Q7
from qiskit import QuantumCircuit
from qiskit.visualization import plot_bloch_vector

# Create 1-qubit circuit
qc = QuantumCircuit(1)

# Apply Hadamard gate
qc.h(0)

# Apply Pauli-Z gate
qc.z(0)

# Display circuit
print(qc)

# Bloch vector for |-> state
# |-> lies on the -X axis
fig = plot_bloch_vector([-1, 0, 0])

# Save Bloch Sphere
fig.savefig("minus_state_bloch_sphere.png",
            dpi=300,
            bbox_inches="tight")

print("Bloch Sphere saved as minus_state_bloch_sphere.png")
```

```python
# Q10
from qiskit import QuantumCircuit
from qiskit.quantum_info import Statevector
from qiskit.visualization import plot_bloch_vector

# Create a single-qubit circuit
qc = QuantumCircuit(1)

# Apply Hadamard gate
qc.h(0)

# Display the circuit
print("Quantum Circuit:")
print(qc)

# Generate the state vector
state = Statevector.from_instruction(qc)

print("\nState Vector:")
print(state)

# Bloch vector for |+> state
fig = plot_bloch_vector([1, 0, 0])

# Save Bloch Sphere
fig.savefig("statevector_bloch_sphere.png",
            dpi=300,
            bbox_inches="tight")

print("\nBloch Sphere saved as statevector_bloch_sphere.png")
```
